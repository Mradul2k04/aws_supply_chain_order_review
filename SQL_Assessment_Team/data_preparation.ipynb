```python
import pandas as pd
import numpy as np
```

```python
df = pd.read_csv(r"D:\AWS_Supply_Chain_Review\SQL_Assessment_Team\dataset\aws_supply_chain_orders_raw.csv", encoding="utf-8")
```

```python
df.head()
```

Task 2: Data Cleaning


Perform:
Missing Value Handling

Replace missing numerical values


```python
df['delivery_time_days']=df['delivery_time_days'].fillna(df['delivery_time_days'].mean())
```

Replace missing categorical values

```python
df['region']=df['region'].fillna('East')
```

Duplicate Removal



Remove duplicate records.

```python
df=df.drop_duplicates()
```

Data Type Standardization


Examples:
Date Columns → DATE/TIMESTAMP
Quantity → INTEGER
Revenue → DECIMAL


```python
df.dtypes
```

```python
df['delivery_time_days']=df['delivery_time_days'].astype(int)
df['delivery_date']=pd.to_datetime(df['delivery_date'])
df['order_date']=pd.to_datetime(df['order_date'])
```

Data Validation


Identify:
Negative quantities

```python
negative_quantities=df[df['order_qty']<0]
negative_quantities
```

Invalid dates

```python
invalid__order_date=df['order_date'].isna()
invalid_delivery_date=df['delivery_date'].isna()
print(invalid__order_date.sum())
print(invalid_delivery_date.sum())
```

Incorrect category mappings

```python
valid_region=['East','West','North','South']
invalid_region=df[df['region'].isin(valid_region)]
invalid_region


valid_status=['Delivered','Delayed','Pending']
invalid_status=df[df['status'].isin(valid_status)]
invalid_status
```

Generate csv file for the cleaned data

```python
df.to_csv(
    "cleaned_data\\aws_supply_chain_orders_clean.csv",
    index=False
)
```
